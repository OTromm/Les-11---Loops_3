## While Loop
In een `While-loop` wordt code herhaald totdat er een vergelijking niet meer waar (`True`) is. De vergelijking wordt gecontroleerd voordat de `While-loop` voor de eerste keer uitgevoerd wordt, en daarna na het uitvoeren van alle code in de `While-loop`:

```python
aantal = 10
while(aantal > 0):
    print(aantal)
```

We maken hier een variabelen `aantal` aan en geven deze de waarde `10`. Vervolgens, net als bij een `If-statement` wordt er een vergelijking gecheckt. Als deze vergelijking `True` is, dan wordt de code in de loop uitgevoerd. In dit geval is `10 > 0` `True`, dus de code in de loop wordt uitgevoerd en het aantal wordt geprint. Daarna wordt de verglijking weer gecheckt of deze nog klopt. In dit geval klopt dit nog, want de waarde van `aantal` veranderdt nooit. We hebben nu dus een `infinite` (oneindige) loop gemaakt.

Om ervoor te zorgen dat we de loop stoppen, moeten we de waarde van `aantal` in de loop kleiner maken zodat de vergelijking `aantal > 0` niet meer klopt. Dit gebeurt dus zodra `aantal` de waarde van 0 wordt.

```python
aantal = 10
while(aantal > 0):
    print(aantal)
    aantal -= 1
```

We hebben nu de regel `aantal -= 1` toegevoegd. Elke keer als de code in de loop wordt uitgevoerd, wordt aantal met 1 verlaagd. Na 10 keer zal de waarden van `aantal` `0` bereiken. Dan wordt de vergelijking nogmaals gecontroleerd en deze keer is de uitkomst van `0 > 0` `False`, dus we stoppen met de loop.
